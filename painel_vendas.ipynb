# Painel de Vendas

Atividade completa de análise e visualização de dados com Pandas, NumPy, Matplotlib e Seaborn.

## Etapa 1 — Preparação e Importação de Dados

### 1) Importação das bibliotecas

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
sns.set_palette("deep")
os.makedirs("graficos", exist_ok=True)
```

### 2) Carregamento do CSV

```python
df = pd.read_csv("vendas.csv")
df["data"] = pd.to_datetime(df["data"])
display(df.head())
```

### 3) Criar a coluna faturamento

```python
df["faturamento"] = df["quantidade"] * df["preco_unitario"]
display(df.head())
```

### 4) Verificar a integridade

```python
df.info()
print("\nValores ausentes:")
display(df.isnull().sum())
print("\nResumo estatístico:")
display(df.describe())
```

### 5) Tratar valores nulos

```python
df["quantidade"] = df["quantidade"].fillna(df["quantidade"].mean())
df["preco_unitario"] = df["preco_unitario"].fillna(df["preco_unitario"].mean())
df["faturamento"] = df["quantidade"] * df["preco_unitario"]
display(df.isnull().sum())
```

## Etapa 2 — Análise Exploratória e Estatísticas

### 6) Faturamento por loja

```python
faturamento_por_loja = df.groupby("loja")["faturamento"].sum().sort_values(ascending=False)
display(faturamento_por_loja.to_frame("faturamento_total"))
```

### 7) Ticket médio de cada loja

```python
ticket_medio = df.groupby("loja")["faturamento"].mean().sort_values(ascending=False)
display(ticket_medio.to_frame("ticket_medio"))
```

### 8) Cinco produtos mais vendidos

```python
top5_produtos = df.groupby("produto")["quantidade"].sum().sort_values(ascending=False).head(5)
display(top5_produtos.to_frame("quantidade_vendida"))
```

### 9) Mês de maior e menor faturamento

```python
faturamento_mensal = df.groupby(df["data"].dt.to_period("M"))["faturamento"].sum()
mes_maior, mes_menor = faturamento_mensal.idxmax(), faturamento_mensal.idxmin()
print(f"Maior: {mes_maior.strftime('%m/%Y')} — R$ {faturamento_mensal.loc[mes_maior]:,.2f}")
print(f"Menor: {mes_menor.strftime('%m/%Y')} — R$ {faturamento_mensal.loc[mes_menor]:,.2f}")
```

### 10) Criar coluna mês

```python
df["mes"] = df["data"].dt.month
display(df[["data","mes"]].head())
```

## Etapa 3 — Visualizações

### 11) Gráfico de barras — Faturamento por loja

```python
plt.figure(figsize=(10,6))
faturamento_por_loja.plot(kind="bar")
plt.title("Faturamento Total por Loja")
plt.xlabel("Loja"); plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=30); plt.tight_layout()
plt.savefig("graficos/faturamento_por_loja.png", dpi=150)
plt.show()
```

### 12) Gráfico de linhas — Evolução mensal

```python
serie = faturamento_mensal.sort_index()
plt.figure(figsize=(10,6))
plt.plot(serie.index.astype(str), serie.values, marker="o")
for x,y in zip(serie.index.astype(str), serie.values):
    plt.annotate(f"R$ {y:,.0f}", (x,y), xytext=(0,8), textcoords="offset points", ha="center", fontsize=8)
plt.title("Evolução Mensal do Faturamento")
plt.xlabel("Mês/Ano"); plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=30); plt.tight_layout()
plt.savefig("graficos/evolucao_mensal.png", dpi=150)
plt.show()
```

### 13) Gráfico de pizza — Faturamento por região

```python
faturamento_por_regiao = df.groupby("regiao")["faturamento"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,8))
plt.pie(faturamento_por_regiao.values, labels=faturamento_por_regiao.index, autopct="%1.1f%%", startangle=90)
plt.title("Distribuição do Faturamento por Região")
plt.tight_layout()
plt.savefig("graficos/faturamento_por_regiao.png", dpi=150)
plt.show()
```

### 14) Scatter — Quantidade x Faturamento

```python
plt.figure(figsize=(10,6))
sns.scatterplot(data=df, x="quantidade", y="faturamento", hue="regiao", style="loja", s=80)
plt.title("Relação entre Quantidade de Itens e Faturamento")
plt.xlabel("Quantidade de Itens Vendidos"); plt.ylabel("Faturamento (R$)")
plt.tight_layout()
plt.savefig("graficos/quantidade_vs_faturamento.png", dpi=150)
plt.show()
```

### 15) Boxplot — Preços por produto

```python
plt.figure(figsize=(11,6))
sns.boxplot(data=df, x="produto", y="preco_unitario", hue="produto", legend=False)
plt.title("Distribuição de Preços Unitários por Produto")
plt.xlabel("Produto"); plt.ylabel("Preço Unitário (R$)")
plt.xticks(rotation=30, ha="right"); plt.tight_layout()
plt.savefig("graficos/boxplot_precos.png", dpi=150)
plt.show()
```

### 16) Heatmap — Correlação

```python
correlacao = df[["quantidade","preco_unitario","faturamento"]].corr()
plt.figure(figsize=(8,6))
sns.heatmap(correlacao, annot=True, cmap="coolwarm", fmt=".2f")
plt.title("Matriz de Correlação das Variáveis Numéricas")
plt.tight_layout()
plt.savefig("graficos/heatmap_correlacao.png", dpi=150)
plt.show()
```

## Etapa 4 — Estilização e Apresentação

### 17) Tema global

```python
sns.set_style("whitegrid")
sns.set_palette("deep")
print("Tema aplicado: whitegrid | Paleta: deep")
```

### 18) Rótulos e títulos

```python
plt.figure(figsize=(10,6))
sns.barplot(x=faturamento_por_loja.index, y=faturamento_por_loja.values)
plt.title("Faturamento Total por Loja")
plt.xlabel("Loja"); plt.ylabel("Faturamento (R$)")
plt.xticks(rotation=30); plt.tight_layout()
plt.savefig("graficos/faturamento_por_loja_revisado.png", dpi=150)
plt.show()
```

### 19) Layout com múltiplos gráficos

```python
fig, axes = plt.subplots(1,2, figsize=(15,6))
sns.barplot(x=faturamento_por_loja.index, y=faturamento_por_loja.values, ax=axes[0])
axes[0].set_title("Faturamento por Loja"); axes[0].set_xlabel("Loja"); axes[0].set_ylabel("Faturamento (R$)")
axes[0].tick_params(axis="x", rotation=30)
sns.barplot(x=faturamento_por_regiao.index, y=faturamento_por_regiao.values, ax=axes[1])
axes[1].set_title("Faturamento por Região"); axes[1].set_xlabel("Região"); axes[1].set_ylabel("Faturamento (R$)")
axes[1].tick_params(axis="x", rotation=30)
plt.suptitle("Comparação de Faturamento por Loja e por Região")
plt.tight_layout()
plt.savefig("graficos/layout_comparativo.png", dpi=150)
plt.show()
```

## Etapa 5 — Exportação e Relatório Final

### 20) Resumo analítico

```python
resumo_analitico = pd.DataFrame({
    "indicador": [
        "Faturamento total", "Loja com maior faturamento", "Maior faturamento mensal",
        "Menor faturamento mensal", "Produto mais vendido", "Região com maior faturamento"
    ],
    "resultado": [
        f"R$ {df['faturamento'].sum():,.2f}",
        faturamento_por_loja.idxmax(),
        f"{mes_maior.strftime('%m/%Y')} — R$ {faturamento_mensal.loc[mes_maior]:,.2f}",
        f"{mes_menor.strftime('%m/%Y')} — R$ {faturamento_mensal.loc[mes_menor]:,.2f}",
        top5_produtos.index[0],
        faturamento_por_regiao.idxmax()
    ]
})
resumo_analitico.to_csv("resumo_analitico.csv", index=False)
display(resumo_analitico)
print("\nPrincipais observações:")
for _, row in resumo_analitico.iterrows():
    print(f"- {row['indicador']}: {row['resultado']}")
```
